```python
import cv2
import numpy as np
import os

class Line:
    def __init__(self):
        self.recent_xfitted = []
        self.best_fit = None
        self.current_fit = [np.array([False])]
        self.radius_of_curvature = None
        self.allx = None
        self.ally = None
        self.detected = False
        
    def add_fit(self, fit):
        self.current_fit = fit
        self.detected = True

class Lane:
    def __init__(self):
        self.left_line = Line()
        self.right_line = Line()

    def update_lane(self, left_fit, right_fit):
        self.left_line.add_fit(left_fit)
        self.right_line.add_fit(right_fit)

def preprocess_image(img):
    # Convert to HSV
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

    # Threshold to get yellow and white colors
    lower_yellow = np.array([10, 100, 100])
    upper_yellow = np.array([30, 255, 255])
    lower_white = np.array([0, 0, 200])
    upper_white = np.array([180, 25, 255])
    
    mask_yellow = cv2.inRange(hsv, lower_yellow, upper_yellow)
    mask_white = cv2.inRange(hsv, lower_white, upper_white)
    
    combined_mask = cv2.bitwise_or(mask_yellow, mask_white)
    return combined_mask

def detect_edges(mask):
    # Apply Canny edge detection
    return cv2.Canny(mask, 50, 150)

def hough_transform(edges):
    # Hough Transform to find lane lines
    return cv2.HoughLinesP(edges, 1, np.pi/180, threshold=50, minLineLength=40, maxLineGap=25)

def process_frame(frame, lane):
    mask = preprocess_image(frame)
    edges = detect_edges(mask)
    lines = hough_transform(edges)

    if lines is not None:
        left_lines = []
        right_lines = []
        
        for line in lines:
            for x1, y1, x2, y2 in line:
                slope = (y2 - y1) / (x2 - x1) if (x2 - x1) != 0 else 0
                if slope < 0:  # Left lane line
                    left_lines.append((x1, y1, x2, y2))
                else:  # Right lane line
                    right_lines.append((x1, y1, x2, y2))

        # Fit a line to the points
        if left_lines:
            left_fit = np.polyfit([x1 for x1, _, x2, _ in left_lines], [y1 for _, y1, _, _ in left_lines], 1)
            lane.left_line.add_fit(left_fit)
        
        if right_lines:
            right_fit = np.polyfit([x1 for x1, _, x2, _ in right_lines], [y1 for _, y1, _, _ in right_lines], 1)
            lane.right_line.add_fit(right_fit)

        # Draw the lines on the image
        draw_lines(frame, left_lines)
        draw_lines(frame, right_lines)
    
    return frame

def draw_lines(img, lines):
    for line in lines:
        # Each line is a tuple of (x1, y1, x2, y2)
        x1, y1, x2, y2 = line
        cv2.line(img, (x1, y1), (x2, y2), (0, 255, 0), 3)
    return img

def main():
    lane = Lane()
    cwd = os.getcwd()
    path = os.path.join(cwd, 'Data', '4695859-uhd_3840_2160_30fps.mp4')
    cap = cv2.VideoCapture(path)
    
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        resize_ = cv2.resize(frame, (1200, 800), interpolation=cv2.INTER_LINEAR)
        processed_frame = process_frame(resize_, lane)
        cv2.imshow('Lane Detection', processed_frame)
        
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
    
    cap.release()
    cv2.destroyAllWindows()

if __name__ == "__main__":
    main()
```
